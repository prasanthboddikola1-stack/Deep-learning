```python
import numpy as np

# Input (4 samples, 2 features)
X = np.array([[0,0],[0,1],[1,0],[1,1]])

# Target (XOR)
Y = np.array([[0],[1],[1],[0]])

# Weights
W1 = np.random.randn(2,3)
W2 = np.random.randn(3,1)

# Activation
def sigmoid(x):
    return 1/(1+np.exp(-x))

# Derivative
def dsigmoid(x):
    return x*(1-x)

lr = 0.1

for epoch in range(1000):

    # Forward
    H = sigmoid(X @ W1)
    O = sigmoid(H @ W2)

    # Error
    error = Y - O

    # Backward
    dO = error * dsigmoid(O)
    dH = (dO @ W2.T) * dsigmoid(H)

    # Update
    W2 += lr * H.T @ dO
    W1 += lr * X.T @ dH

print(O)
```

```python
import torch

torch.manual_seed(42)

# Xavier Initialization
def xavier(in_features, out_features):
    return torch.randn(in_features, out_features) * (2 / (in_features + out_features)) ** 0.5

# He Initialization
def he(in_features, out_features):
    return torch.randn(in_features, out_features) * (2 / in_features) ** 0.5

# Input
x = torch.randn(32, 64)

# Choose Initialization
W = he(64, 64)          # Change to xavier(64, 64) for Xavier initialization

# 20-layer MLP
for i in range(20):
    x = torch.relu(x @ W)

print("Final Output Shape:", x.shape)
print("Mean:", x.mean().item())
print("Variance:", x.var().item())
```

```python
import numpy as np

# Input Image (5x5)
image = np.array([
    [1, 2, 3, 0, 1],
    [4, 5, 6, 1, 2],
    [7, 8, 9, 2, 3],
    [1, 2, 3, 4, 5],
    [6, 7, 8, 9, 0]
])

# 3x3 Filter
kernel = np.array([
    [1, 0, -1],
    [1, 0, -1],
    [1, 0, -1]
])

# Output Size
h, w = image.shape
k = kernel.shape[0]

output = np.zeros((h - k + 1, w - k + 1))

# Convolution
for i in range(h - k + 1):
    for j in range(w - k + 1):
        region = image[i:i+k, j:j+k]
        output[i, j] = np.sum(region * kernel)

print("Output:")
print(output)
```

```python
import torch

torch.manual_seed(42)

# Parameter to optimize
w = torch.tensor([5.0])

# Gradient
grad = torch.tensor([2.0])

# Adam parameters
lr = 0.01
beta1 = 0.9
beta2 = 0.999
eps = 1e-8

m = torch.zeros_like(w)
v = torch.zeros_like(w)

for t in range(1, 11):

    # Update moments
    m = beta1 * m + (1 - beta1) * grad
    v = beta2 * v + (1 - beta2) * grad**2

    # Bias correction
    m_hat = m / (1 - beta1**t)
    v_hat = v / (1 - beta2**t)

    # Adam update
    w = w - lr * m_hat / (torch.sqrt(v_hat) + eps)

    print(f"Step {t}: {w.item():.4f}")
```

```python
import torch

torch.manual_seed(42)

# Input (4 samples, 3 features)
x = torch.randn(4, 3)

# Learnable parameters
gamma = torch.ones(3)
beta = torch.zeros(3)

# Mean and Variance
mean = x.mean(dim=0)
var = x.var(dim=0, unbiased=False)

# Normalize
x_hat = (x - mean) / torch.sqrt(var + 1e-5)

# Batch Normalization
y = gamma * x_hat + beta

print("Input:\n", x)
print("\nNormalized Output:\n", y)
```

```python
import torch

torch.manual_seed(42)

# Input sequence (5 time steps, 3 features)
x = torch.randn(5, 3)

# Hidden size
hidden_size = 4

# Forward hidden state
h_f = torch.zeros(hidden_size)

# Backward hidden state
h_b = torch.zeros(hidden_size)

# Weights
W = torch.randn(3, hidden_size)

# Forward pass
forward_output = []
for t in range(len(x)):
    h_f = torch.tanh(x[t] @ W + h_f)
    forward_output.append(h_f)

# Backward pass
backward_output = []
for t in reversed(range(len(x))):
    h_b = torch.tanh(x[t] @ W + h_b)
    backward_output.insert(0, h_b)

# Combine forward and backward outputs
bilstm_output = torch.cat(
    (torch.stack(forward_output),
     torch.stack(backward_output)),
    dim=1
)

print(bilstm_output)
```

```python
import torch
import torch.nn as nn

class ResidualBlock(nn.Module):
    def __init__(self):
        super().__init__()

        self.conv1 = nn.Conv2d(3, 3, kernel_size=3, padding=1)
        self.relu = nn.ReLU()
        self.conv2 = nn.Conv2d(3, 3, kernel_size=3, padding=1)

    def forward(self, x):
        identity = x
        out = self.conv1(x)
        out = self.relu(out)
        out = self.conv2(out)
        out = out + identity      # Skip Connection
        out = self.relu(out)
        return out

# Input
x = torch.randn(1, 3, 32, 32)

# Model
model = ResidualBlock()

# Output
y = model(x)

print("Output Shape:", y.shape)
```

```python
import torch
import torch.nn as nn

class SimpleUNet(nn.Module):
    def __init__(self):
        super().__init__()

        self.encoder = nn.Conv2d(3, 16, kernel_size=3, padding=1)
        self.decoder = nn.Conv2d(16, 1, kernel_size=3, padding=1)
        self.relu = nn.ReLU()
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x = self.relu(self.encoder(x))
        x = self.sigmoid(self.decoder(x))
        return x

# Input Image
x = torch.randn(1, 3, 128, 128)

# Model
model = SimpleUNet()

# Output Mask
output = model(x)

print("Segmentation Mask Shape:", output.shape)
```

```python
import torch
import torch.nn.functional as F

# Predicted box (x1, y1, x2, y2)
pred_box = torch.tensor([10., 10., 30., 30.])

# Ground truth box
true_box = torch.tensor([15., 15., 35., 35.])

# IoU Function
def iou(box1, box2):
    x1 = max(box1[0], box2[0])
    y1 = max(box1[1], box2[1])
    x2 = min(box1[2], box2[2])
    y2 = min(box1[3], box2[3])

    inter = max(0, x2-x1) * max(0, y2-y1)
    area1 = (box1[2]-box1[0]) * (box1[3]-box1[1])
    area2 = (box2[2]-box2[0]) * (box2[3]-box2[1])

    return inter / (area1 + area2 - inter)

# CIoU Loss (Simplified)
ciou_loss = 1 - iou(pred_box, true_box)

# Classification Loss
pred_class = torch.tensor([[2.0, 0.5]])
true_class = torch.tensor([0])

cls_loss = F.cross_entropy(pred_class, true_class)

print("CIoU Loss:", ciou_loss.item())
print("Classification Loss:", cls_loss.item())
```

```python
import torch
import torch.nn as nn

# Generator
class Generator(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc = nn.Linear(100, 28 * 28)

    def forward(self, z):
        return self.fc(z)

# Critic
class Critic(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc = nn.Linear(28 * 28, 1)

    def forward(self, x):
        return self.fc(x)

G = Generator()
C = Critic()

# Random Noise
z = torch.randn(4, 100)

# Fake Image
fake = G(z)

# Critic Score
score = C(fake)

# Wasserstein Loss
loss = -score.mean()

print("WGAN Loss:", loss.item())
```

```python
import torch
import torch.nn as nn

# Input Image (1 image, 3 channels, 32x32)
x = torch.randn(1, 3, 32, 32)

# Patch Embedding
patch_embed = nn.Conv2d(3, 64, kernel_size=4, stride=4)

# Convert image into patches
patches = patch_embed(x)

# Flatten patches
patches = patches.flatten(2).transpose(1, 2)

# Multi-Head Self Attention
attention = nn.MultiheadAttention(embed_dim=64, num_heads=4, batch_first=True)

output, _ = attention(patches, patches, patches)

print("Output Shape:", output.shape)
```

```python
import torch
import torch.nn as nn

# Convolutional Autoencoder
class AutoEncoder(nn.Module):
    def __init__(self):
        super().__init__()

        # Encoder
        self.encoder = nn.Sequential(
            nn.Conv2d(1, 8, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )

        # Decoder
        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(8, 1, 2, stride=2),
            nn.Sigmoid()
        )

    def forward(self, x):
        x = self.encoder(x)
        x = self.decoder(x)
        return x

# Input Image
x = torch.randn(1, 1, 28, 28)

# Model
model = AutoEncoder()

# Reconstructed Image
output = model(x)

print("Output Shape:", output.shape)
```

```python
import torch
import torch.nn as nn

# Simple TCN
class TCN(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1 = nn.Conv1d(1, 8, kernel_size=3, padding=2, dilation=1)
        self.conv2 = nn.Conv1d(8, 16, kernel_size=3, padding=4, dilation=2)
        self.relu = nn.ReLU()

    def forward(self, x):
        x = self.relu(self.conv1(x))
        x = self.relu(self.conv2(x))
        return x

# Input (Batch, Channels, Sequence Length)
x = torch.randn(1, 1, 20)

# Model
model = TCN()

# Output
output = model(x)

print("Output Shape:", output.shape)
```

```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
import torch.distributed as dist
from torch.nn.parallel import DistributedDataParallel as DDP

def main():
    # Set environment variables for single-process distributed training simulation.
    # These are typically set by a distributed launcher (e.g., torch.run).
    # For a single process in Colab, we simulate rank 0 out of 1 total process.
    os.environ['MASTER_ADDR'] = 'localhost'  # Required for GLOO backend
    os.environ['MASTER_PORT'] = '29500'      # Required for GLOO backend, pick an unused port
    os.environ['RANK'] = '0'                 # This process is rank 0
    os.environ['WORLD_SIZE'] = '1'           # Total number of processes is 1
    os.environ['LOCAL_RANK'] = '0'           # Local rank for this process is 0

    # Initialize the process group
    dist.init_process_group(backend="gloo", rank=int(os.environ['RANK']), world_size=int(os.environ['WORLD_SIZE']))

    local_rank = int(os.environ["LOCAL_RANK"])
    # For simplicity, using CPU. In a multi-GPU setup, you'd assign a specific GPU.
    device = torch.device("cpu")

    model = nn.Linear(10, 2).to(device)
    # When world_size is 1, DDP wraps the model but doesn't perform actual distribution.
    model = DDP(model)

    optimizer = optim.SGD(model.parameters(), lr=0.01)
    criterion = nn.MSELoss()

    x = torch.randn(8, 10).to(device)
    y = torch.randn(8, 2).to(device)

    optimizer.zero_grad()

    output = model(x)

    loss = criterion(output, y)

    loss.backward()

    optimizer.step()

    print("Loss:", loss.item())

    # Destroy the process group at the end
    dist.destroy_process_group()

if __name__ == "__main__":
    main()
```
